## Q1. Basic Line Plot

Create a line plot using:

```python
days = [1, 2, 3, 4, 5]
sales = [20, 35, 30, 45, 50]
```

Plot `days` on the X-axis and `sales` on the Y-axis.

**Expected Output:** A simple line graph showing sales for 5 days.

````python
# Write your answer here
days = [1, 2, 3, 4, 5]
sales = [20, 35, 30, 45, 50]

plt.figure(figsize=(8, 6))
plt.plot(days, sales, marker = 'o')
plt.xlabel('Days')
plt.ylabel('Sales')
plt.title('Sales Over Time')
plt.grid(True)
plt.show()
````

## Q2. Figure

Using the same data, create a figure explicitly using `plt.figure()` before plotting.

**Expected Output:** The same line graph created inside an explicitly created figure.

````python
# Write your answer here
plt.figure(figsize=(8, 6), dpi=200)
plt.plot(days, sales, marker = 'o')
plt.xlabel('Days')
plt.ylabel('Sales')
plt.title('Sales Over Time')
plt.grid(True)
plt.show()
````

## Q3. Figure Size

Create a line plot using:

```python

```

**Expected Output:** A wider figure containing the revenue line plot.

## Q7. Legend

Plot two lines on the same graph:

```python
x = [1, 2, 3, 4, 5]
sales_a = [10, 20, 25, 30, 40]
sales_b = [15, 18, 28, 35, 45]
```

Add a legend for both lines.

**Expected Output:** Two lines with a legend identifying each line.

````python
# Write your answer here
x = [1, 2, 3, 4, 5]
sales_a = [10, 20, 25, 30, 40]
sales_b = [15, 18, 28, 35, 45]

plt.figure(figsize=(8, 6))
plt.plot(x, sales_a, label='Sales A', marker='o', linestyle='--')
plt.plot(x, sales_b, label='Sales B', marker='o', linestyle='--')
plt.xlabel('Day')
plt.ylabel('Sales')
plt.title('Sales Over Time')
plt.grid(True)
plt.legend()
plt.show()
````

## Q9. Subplot

Use `plt.subplot()` to create **2 plots side by side**.

- First plot: `[10, 20, 30, 40]`
- Second plot: `[15, 25, 20, 35]`

Give both plots suitable titles.

**Expected Output:** Two plots in one figure.

````python
# Write your answer here
x = [10, 20, 30, 40]
y = [15, 25, 20, 35]

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(x)
plt.title('First Plot')

plt.subplot(1, 2, 2)
plt.plot(y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Second Plot')
plt.grid(True)
plt.show()
````

## Q11. Figure + Subplots

Create a figure with size **10 × 6** and use `plt.subplot()` to create two plots:

- Left: line plot of `sales = [20, 30, 25, 40, 50]`
- Right: line plot of `profit = [5, 8, 7, 12, 15]`

Give both plots suitable titles.

**Expected Output:** Two plots inside one 10 × 6 figure.

````python
# Write your answer here
sales = [20, 30, 25, 40, 50]
profit = [5, 8, 7, 12, 15]

plt.figure(figsize=(10, 6))
plt.subplot(1, 2, 1)
plt.plot(sales)
plt.title('Sales')

plt.subplot(1, 2, 2)
plt.plot(profit)
plt.title('Profit')

plt.tight_layout()
plt.show()
````
